```python
def score_level(score):
    #对于超出0〜100范围的数值，抛出ValueError。
    if score < 0 or score > 100:
        raise ValueError("成绩必须在0到100之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 70:
        return "中等"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试调用
for s in [95, 85, 75, 60, 59, 100, 0]:
    print(f"{s:>3} 分 -> {score_level(s)}")

try:
    score_level(120)
except ValueError as e:
    print("捕获到异常:", e)
```

```python
#分别使⽤for和while计算1到100的整数和，⽐较两种实现的结果。

total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   求和结果:", total_for)
print("while 求和结果:", total_while)
```

```python
# 任务3：递归版与迭代版阶乘（0! = 1），比较 n = 0, 1, 5, 10
def factorial_recursive(n):
    if n < 0:
        raise ValueError("阶乘要求非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    if n < 0:
        raise ValueError("阶乘要求非负整数")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in [0, 1, 5, 10]:
    r1, r2 = factorial_recursive(n), factorial_iterative(n)
    print(f"{n:>2}! 递归 = {r1:<12} 迭代 = {r2:<12} 一致: {r1 == r2}")
```

```python
# 任务4：summary(values) 以元组返回 (最小值, 最大值, 平均值)，并用元组解包接收
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

low, high, avg = summary([85, 58, 92, 76, 60, 89])
print("最小值:", low)
print("最大值:", high)
print("平均值:", round(avg, 2))
```

```python
# 任务1：查看工作目录，用 os.path.join() 构造路径，创建数据目录
import os

print("当前工作目录:", os.getcwd())

data_dir = os.path.join("exp01_data")            # 相对路径
os.makedirs(data_dir, exist_ok=True)             # 已存在也不报错
students_path = os.path.join(data_dir, "students.txt")

print("数据目录:", os.path.abspath(data_dir))
print("数据文件:", os.path.abspath(students_path))
```

```python
# 任务2：写入 UTF-8 文本文件，再读取（跳过表头，用字符串方法解析）
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(students_path, "w", encoding="utf-8") as f:
    f.write(content)

with open(students_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("总行数:", len(lines))
print("表头:", lines[0].strip())

for line in lines[1:]:                 # 跳过表头
    parts = line.strip().split(",")
    if len(parts) == 2:
        print(f"  {parts[0]:<8} -> {parts[1]}")
    else:
        print(f"  [字段数异常] {line.strip()!r}")
```

```python
# 任务3：parse_student(line) —— 字段数、姓名非空、成绩 0~100 校验，无效抛 ValueError
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数应为 2，实际为 {len(parts)}")
    name = parts[0].strip()
    score_str = parts[1].strip()
    if name == "":
        raise ValueError("姓名不能为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数: {score_str!r}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围: {score}")
    return {"name": name, "score": score}

print(parse_student("Alice,85"))
print(parse_student(" Bob , 90 "))
```

```python
# 任务4：逐行处理，捕获无效记录（记录行号与原因），继续处理后续记录
valid_records = []
invalid_records = []

with open(students_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for lineno, line in enumerate(lines[1:], start=2):   # 第2行起为数据行
    text = line.strip()
    if text == "":
        continue
    try:
        valid_records.append(parse_student(text))
    except ValueError as e:
        invalid_records.append((lineno, text, str(e)))
        print(f"第 {lineno} 行无效: {text!r} -> {e}")

print()
print("有效记录:")
for r in valid_records:
    print("  ", r)
print("有效记录数:", len(valid_records))
print("无效记录数:", len(invalid_records))
```

```python
# 任务5：统计有效/无效记录数、平均分、及格率写入 report.txt，并演示 w 与 a 模式
n_valid = len(valid_records)
n_invalid = len(invalid_records)
avg = sum(r["score"] for r in valid_records) / n_valid
pass_rate = sum(1 for r in valid_records if r["score"] >= 60) / n_valid

report = (
    f"有效记录数: {n_valid}\n"
    f"无效记录数: {n_invalid}\n"
    f"平均分: {avg:.2f}\n"
    f"及格率: {pass_rate:.1%}\n"
)
report_path = os.path.join(data_dir, "report.txt")

# w 模式：会先清空文件原有内容，再写入
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report)
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# a 模式：在文件末尾追加，不清空已有内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("--- 追加一行（a 模式）---\n")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# 任务6：用 pickle 保存有效记录，再读取并验证与原数据相等
import pickle

pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)

print("恢复的数据:", loaded)
print("与原数据相等:", loaded == valid_records)
```

```python
# 任务1：定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

s = Student("Alice", 85)
print(s)
print("是否及格:", s.is_passed())
```

```python
# 任务2：为实验4中的每条有效记录创建 Student 对象，输出信息与及格状态
students = [Student(r["name"], r["score"]) for r in valid_records]

for st in students:
    flag = "及格" if st.is_passed() else "不及格"
    print(f"{st}:{flag}")
```

```python
# 任务3：定义继承自 Student 的 GraduateStudent
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)          # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")

gs = GraduateStudent("Carol", 92, "机器学习")
print(gs)
print("是否及格:", gs.is_passed())
print("是否为 Student 的实例:", isinstance(gs, Student))
```

```python
# 任务4：直接赋值 / copy.copy() / copy.deepcopy() 对比
import copy

def make_original():
    return {"name": "Alice", "scores": [85, 90]}

#直接赋值
a = make_original()
b = a
b["scores"].append(100)
print("直接赋值：原始:", a, ", copy:", b, ", 同一对象:", a is b)

# 浅拷贝
a = make_original()
c = copy.copy(a)
c["scores"].append(100)
print("浅拷贝：原始:", a, ", copy:", c, ", 同一对象:", a is c)       

# 深拷贝
a = make_original()
d = copy.deepcopy(a)
d["scores"].append(100)
print("深拷贝：原始:", a, ", copy:", d, ", 同一对象:", a is d)
```
