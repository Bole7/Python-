```python
import os, uuid, hashlib, platform, getpass

# ===== 1. 填写学号 =====
STUDENT_ID = 20230651625  # 修改为自己的学号

# ===== 2. 打印 conda 环境 =====
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")

# ===== 3. 机器指纹（绑定当前电脑）=====
MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]

# ===== 4. 生成唯一 ID =====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号]     {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID]   {RUN_ID}")
```

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

for a, b in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({a}, {b}) = {gcd(a, b)}")

print()
print("gcd(-2018, 2019) =", gcd(-2018, 2019))
print("gcd(2018, -2019) =", gcd(2018, -2019))
print("gcd(-8, -12)     =", gcd(-8, -12))

# 当参数含负数时，函数返回值的绝对值仍然正确，但符号不可靠：

# gcd(-2018, 2019) = 1
# gcd(2018, -2019) = -1
# gcd(-8, -12)     = -4

# 原因：Python 中 % 的结果符号跟随除数，循环结束时 m 可能是负数。而数学上最大公约数约定为非负，所以这种行为不符合约定。修正方式：返回前取绝对值return abs(m)
```

```python
import math

# 1. 
def euler_limit(n):
    return (1 + 1 / n) ** n

# 2.
def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1 / fact
    return total

# 3.
def euler_approx(eps):
    total = 0.0
    k = 0
    fact = 1
    while True:
        if k > 0:
            fact *= k
        total += 1 / fact
        if abs(total - math.exp(1)) < eps:
            return total
        k += 1

# 4.
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print("euler_limit(1)     =", euler_limit(1))
print("euler_limit(100)   =", euler_limit(100))
print("euler_limit(10000) =", euler_limit(10000))
print("math.exp(1)        =", math.exp(1))

print()
print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))

print()
print("euler_approx(1e-6) =", euler_approx(1e-6))
print("euler_approx(1e-12) =", euler_approx(1e-12))

print()
print("print_euler_sum_table(6):")
print_euler_sum_table(6)
```

```python
# 1. 
def is_power_of_2(n):
    if n < 1:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1

# 2. 
def is_power(b, n):
    if b < 2:            
        return n == 1
    if n < 1:
        return False
    while n % b == 0:
        n //= b
    return n == 1

for x in [1, 2, 3, 4, 6, 8, 16, 17, 1024, 1000]:
    print(f"is_power_of_2({x}) = {is_power_of_2(x)}")

print()
print("is_power(3, 9)   =", is_power(3, 9))
print("is_power(3, 8)   =", is_power(3, 8))
print("is_power(2, 16)  =", is_power(2, 16))
print("is_power(5, 125) =", is_power(5, 125))
```
