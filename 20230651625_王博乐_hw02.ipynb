```python
def is_palindrome(s):
    # 忽略空格和大小写：去除所有空格并转成小写
    cleaned = ''.join(ch for ch in s.lower() if ch != ' ')
    # 正序与倒序相同则为回文
    return cleaned == cleaned[::-1]

print(is_palindrome("tacocat"), is_palindrome("T A C O cat"), is_palindrome("dog"))
```

```python
def is_abecedarian(s):
    letters = s.replace(" ", "").lower()
    return all(letters[i] <= letters[i + 1] for i in range(len(letters) - 1))

print(is_abecedarian("adder"), is_abecedarian("beet"), is_abecedarian("abcd efgh xyz"), is_abecedarian("dog"))
```

```python
def remove_vowels(s):
    # 移除所有元音字母
    return ''.join(ch for ch in s if ch not in 'aeiouAEIOU  ')
print(remove_vowels("cat"), remove_vowels("audio"), remove_vowels("aeiou"))
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表（list）")
    return lst[::-1]
print(list_reverse([1, 2, 3]))
```

```python
def is_sorted(seq):
    # 只遍历一次：相邻两两比较，一旦出现降序立即返回 False
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True
print(is_sorted([1, 2, 2, 5]), is_sorted([3, 1]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)) or isinstance(s, bool):
        raise TypeError("标量必须是整数或浮点数")
    for x in t:
        if not isinstance(x, (int, float)) or isinstance(x, bool):
            raise TypeError("元组的元素必须都是整数或浮点数")
    return tuple(x * s for x in t)

mx = ((1, 2, 3), (4, 5, 6))
print(vec_scalar_mult((1, 2.5, -3), 2))
```

```python
def vec_inner_product(u, v):
    if not isinstance(u, tuple) or not isinstance(v, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(u) != len(v):
        raise ValueError("两个向量的长度必须一致")
    return float(sum(float(a) * float(b) for a, b in zip(u, v)))
    
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False                        # 空矩阵视为无效
    for row in mx:
        if not isinstance(row, tuple):      # 每行本身必须是元组
            return False
    n = len(mx[0])
    for row in mx:
        if len(row) != n:                   # 各行长度必须相同
            return False
        for x in row:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                return False
    return True

print(check_valid_mx(mx), check_valid_mx(((1, 2), (3,))), check_valid_mx([[1, 2]]))
```

```python
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("第一个参数不是有效的矩阵")
    if not isinstance(v, tuple):
        raise TypeError("第二个参数必须是元组")
    n = len(mx[0])
    if len(v) != n:
        raise ValueError("向量长度应为 " + str(n) + "，实际为 " + str(len(v)))
    return tuple(sum(float(a) * float(b) for a, b in zip(row, v)) for row in mx)

print(mx_vec_mult(mx, (1, 1, 1)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if elements is None:                      # 只给维度：默认全 0
            elements = (0,) * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("元素个数应为 " + str(dim) + "，实际为 " + str(len(elements)))
        for x in elements:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                raise TypeError("元素必须都是整数或浮点数")
        self.dim = dim
        self.coords = tuple(elements)

    def _check(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check(other)
        return self.coords == other.coords

    def __ne__(self, other):
        self._check(other)
        return self.coords != other.coords

    def __lt__(self, other):
        self._check(other)
        return self.coords < other.coords

    def __le__(self, other):
        self._check(other)
        return self.coords <= other.coords

    def __gt__(self, other):
        self._check(other)
        return self.coords > other.coords

    def __ge__(self, other):
        self._check(other)
        return self.coords >= other.coords

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot 的参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积")
        return float(sum(a * b for a, b in zip(self.coords, other.coords)))

    def __mul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dim, tuple(x * other for x in self.coords))
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法逐元素相乘")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.coords, other.coords)))
        raise TypeError("只能与标量或 Vector 相乘")

    __rmul__ = __mul__

    def __repr__(self):
        return "Vector(" + str(self.dim) + ", " + repr(self.coords) + ")"


a, b = Vector(2, (2, 4)), Vector(2, (2, 5))
print(a < b, a == Vector(2, [2, 4]), a.dot(b))
print(2 * a, a * 2, Vector(2, (2, 3)) * Vector(2, (4, 5)))
print(Vector(3))
```
